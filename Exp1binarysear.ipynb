```python
def binary_search(arr, target):
    """
    Classic Binary Search on a sorted array.
    Always probes the midpoint: mid = (lo + hi) // 2
    Best case:  O(1)         — target is midpoint
    Worst case: O(log n)     — consistent regardless of distribution
    """
    lo, hi = 0, len(arr) - 1
    probes = 0
    trace = []

    while lo <= hi:
        mid = (lo + hi) // 2
        probes += 1

        if arr[mid] == target:
            trace.append((probes, mid, arr[mid], "✓ FOUND"))
            return mid, probes, trace
        elif arr[mid] < target:
            trace.append((probes, mid, arr[mid], "go RIGHT ▶"))
            lo = mid + 1
        else:
            trace.append((probes, mid, arr[mid], "go LEFT ◀"))
            hi = mid - 1

    trace.append((probes + 1, -1, -1, "✗ NOT FOUND"))
    return -1, probes, trace
```
