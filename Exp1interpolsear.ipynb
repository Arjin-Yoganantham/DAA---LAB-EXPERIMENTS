```python
def interpolation_search(arr, target):
    """
    Interpolation Search on a sorted array.
    Uses position formula: pos = lo + ((target - arr[lo]) * (hi - lo))
                                         / (arr[hi] - arr[lo])
    Best case:  O(log log n)  — uniformly distributed data
    Worst case: O(n)          — skewed/non-uniform data
    """
    lo, hi = 0, len(arr) - 1
    probes = 0
    trace = []

    while lo <= hi and arr[lo] <= target <= arr[hi]:
        if lo == hi:
            probes += 1
            trace.append((probes, lo, arr[lo], "FOUND" if arr[lo] == target else "NOT FOUND"))
            if arr[lo] == target:
                return lo, probes, trace
            break

        # Interpolation formula — key difference from binary search
        pos = lo + ((target - arr[lo]) * (hi - lo)) // (arr[hi] - arr[lo])
        probes += 1

        if arr[pos] == target:
            trace.append((probes, pos, arr[pos], "✓ FOUND"))
            return pos, probes, trace
        elif arr[pos] < target:
            trace.append((probes, pos, arr[pos], "go RIGHT ▶"))
            lo = pos + 1
        else:
            trace.append((probes, pos, arr[pos], "go LEFT ◀"))
            hi = pos - 1

    return -1, probes, trace  # not found
```
